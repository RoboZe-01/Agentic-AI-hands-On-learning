## Importing dependencies

```python
import os      # TO import test image files
import cv2     # To work on images
from PIL import Image  # Image submodule to work with pil images
import pytesseract as pt  # Python wrapper for tesseract
```

```python
test_img_path = "../data/test images/"
create_path = lambda f : os.path.join(test_img_path, f)

test_image_files = os.listdir(test_img_path)

for f in test_image_files:
    print(f)
```

```python
def show_image(img_path , size = (500,500)):
    image = cv2.imread(img_path)
    image = cv2.resize(image,size)
    cv2.imshow("IMAGE",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
